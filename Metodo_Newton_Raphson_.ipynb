```python
"""Resuelve x = exp(-x) usando el método de Newton-Raphson."""
from math import exp, isfinite


def f(x):
    """Función que debe hacerse cero."""
    return x - exp(-x)


def derivada_f(x):
    """Derivada analítica de f."""
    return 1 + exp(-x)


def newton_raphson(x0=0.5, tolerancia=0.001, max_iteraciones=100):
    """Busca la raíz comparando dos aproximaciones consecutivas."""
    if tolerancia <= 0:
        raise ValueError("La tolerancia debe ser positiva.")

    x_actual = x0
    print("Método de Newton-Raphson para x = e^(-x)")
    print(f"Valor inicial: {x0:.6f} | Tolerancia: {tolerancia}")
    print(f"{'i':>3} {'x_i':>14} {'f(x_i)':>14} {'f\u2032(x_i)':>14} {'error':>14}")
    print(f"{0:>3} {x_actual:>14.9f} {f(x_actual):>14.9f} "
          f"{derivada_f(x_actual):>14.9f} {'---':>14}")

    for i in range(1, max_iteraciones + 1):
        pendiente = derivada_f(x_actual)
        if not isfinite(pendiente) or abs(pendiente) < 1e-14:
            raise ArithmeticError("La derivada es cero o no es válida.")

        x_siguiente = x_actual - f(x_actual) / pendiente
        error = abs(x_siguiente - x_actual)
        print(f"{i:>3} {x_siguiente:>14.9f} {f(x_siguiente):>14.9f} "
              f"{derivada_f(x_siguiente):>14.9f} {error:>14.9f}")

        if error < tolerancia:
            print(f"\nRaíz aproximada: x = {x_siguiente:.9f}")
            print(f"Verificación: f(x) = {f(x_siguiente):.9e}")
            print(f"Se alcanzó la tolerancia en {i} iteraciones.")
            return x_siguiente

        x_actual = x_siguiente

    raise RuntimeError("No se alcanzó la tolerancia en el límite de iteraciones.")


if __name__ == "__main__":
    newton_raphson()
```
